```python
import pandas as pd, glob
df = pd.concat([pd.read_csv(f, encoding="latin-1", on_bad_lines="skip") for f in glob.glob("data/*.csv")])
df.head()
```

```python
#Which team won the most home games overall?
home_wins = df[df["FTR"] == "H"]
home_wins["HomeTeam"].value_counts()
```

```python
#What's the average number of goals per match?
df["total_goals"] = df["FTHG"] + df["FTAG"] 
avg_goals = df["total_goals"].mean()
avg_goals
```

```python
#How many matches did Arsenal play, and how many did they win?
Arsl_matches = df[(df["HomeTeam"] == "Arsenal") | (df["AwayTeam"] == "Arsenal")]
len(Arsl_matches)
Arsl_home_wins = df[(df["HomeTeam"] == "Arsenal") & (df["FTR"] == "H")]
Arsl_away_wins = df[(df["AwayTeam"] == "Arsenal") & (df["FTR"] == "A")]
total_wins = len(Arsl_home_wins) + len(Arsl_away_wins)
total_wins
```
